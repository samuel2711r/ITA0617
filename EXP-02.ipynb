```python

# Candidate Elimination Algorithm

# Step 1: Define the training dataset
data = [
    ['Sunny', 'Warm', 'Yes'],
    ['Sunny', 'Cold', 'Yes'],
    ['Rainy', 'Warm', 'No'],
    ['Sunny', 'Warm', 'Yes']
]

# Step 2: Initialize S and G
S = data[0][:-1]
G = ['?'] * len(S)

print("Initial S =", S)
print("Initial G =", G)

# Step 3: Process training examples
for x in data:

    if x[-1] == 'Yes':

        # Generalize S for positive examples
        for i in range(len(S)):
            if S[i] != x[i]:
                S[i] = '?'

    else:

        # Specialize G for negative examples
        for i in range(len(S)):
            if S[i] != x[i]:
                G[i] = S[i]

    print("S =", S, "G =", G)

# Step 4: Display final hypotheses
print("\nFinal S =", S)
print("Final G =", G)
```
