```python

from sklearn.tree import DecisionTreeClassifier

# Create dataset
X = [[500], [600], [700], [800]]
y = ['Poor', 'Fair', 'Good', 'Excellent']

# Create Decision Tree model
m = DecisionTreeClassifier(random_state=42)

# Train the model
m.fit(X, y)

# Predict credit score category
print(m.predict([[750]]))
```
