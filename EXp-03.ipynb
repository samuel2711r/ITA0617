```python

# EXPERIMENT 3 - DECISION TREE (ID3) ALGORITHM

# Step 1: Import Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

# Step 2: Create training dataset
X = [[0, 0],
     [0, 1],
     [1, 0],
     [1, 1]]

# Step 3: Define target values
y = ['No', 'No', 'Yes', 'Yes']

# Step 4: Create Decision Tree model using Entropy
clf = DecisionTreeClassifier(criterion='entropy')

# Step 5: Train the model
clf.fit(X, y)

# Step 6: Classify a new sample
prediction = clf.predict([[1, 0]])

# Step 7: Display the result
print("Predicted Result:", prediction)
```
