```python

# EXPERIMENT 4 - ARTIFICIAL NEURAL NETWORK
# USING BACKPROPAGATION

# Step 1: Import MLPClassifier
from sklearn.neural_network import MLPClassifier

# Step 2: Create training dataset
X = [[0, 0],
     [0, 1],
     [1, 0],
     [1, 1]]

# Step 3: Define target values
y = [0, 1, 1, 0]

# Step 4: Create ANN model
ann = MLPClassifier(
    hidden_layer_sizes=(2,),
    max_iter=1000,
    random_state=42
)

# Step 5: Train the model
ann.fit(X, y)

# Step 6: Test using a new sample
prediction = ann.predict([[1, 0]])

# Step 7: Display prediction
print("Prediction:", prediction)
```
