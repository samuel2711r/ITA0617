```python

# FIND-S Algorithm

# Step 1: Define the training dataset
data = [
    ['Sunny', 'Warm', 'Normal', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Yes'],
    ['Rainy', 'Cold', 'High', 'No'],
    ['Sunny', 'Warm', 'High', 'Yes']
]

# Step 2: Initialize hypothesis
h = ['0'] * (len(data[0]) - 1)

print("Initial Hypothesis:", h)

# Step 3: Process training examples
for row in data:

    # Consider only positive examples
    if row[-1] == 'Yes':

        if h[0] == '0':
            h = row[:-1]

        else:
            for i in range(len(h)):
                if h[i] != row[i]:
                    h[i] = '?'

    print("Current Hypothesis:", h)

# Step 4: Display final hypothesis
print("\nFinal Hypothesis:", h)
```
