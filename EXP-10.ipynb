```python

from sklearn.mixture import GaussianMixture
import numpy as np

# Create dataset
X = np.array([[1], [2], [3], [10], [11], [12]])

# Create EM model with 2 clusters
g = GaussianMixture(n_components=2, random_state=42)

# Train the model
g.fit(X)

# Predict cluster labels
print(g.predict(X))
```
